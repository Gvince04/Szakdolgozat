```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
```

```python
preprocessed_path = "zengo_preprocessed/"
```

```python
headset_data = None

for root, dirs, files in os.walk(preprocessed_path):
    for file in files:
        if file == "headset.npy":
            full_path = os.path.join(root, file)
            headset_data = np.load(full_path)
            print("File found and loaded!")
            break
    else:
        continue

    break
```

```python
print(f"Headset data shape: {headset_data.shape}")
```

```python
plt.plot(headset_data[:, 1])
```

```python
headset_df = pd.DataFrame(headset_data)

headset_df.iloc[:10, :5]
```

```python
for i in range(headset_data.shape[1]):
    plt.figure(figsize=(10, 6))
    plt.plot(headset_data[:200, i], label='Sensor data')
    
    plt.grid(True)
    plt.legend()

    plt.show()
```
